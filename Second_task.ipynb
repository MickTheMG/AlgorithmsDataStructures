```python
import numpy as np
from collections import deque
```

```python
class Node:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

### Is the list cycled?

```python
def isCycled(head: Node) -> bool:
    if not head or not head.next:
        return False

    slow = head
    fast = head

    while slow != fast:
        if not fast or not fast.next:
            return False

        slow = slow.next
        fast = fast.next.next

    return True
```

### Reverse linked list

```python
def reverse(head: Node) -> Node:
    prev = None
    current = head

    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node

    # head = prev
    return prev
```

### Search the middle

```python
def middle_search(head: Node) -> Node:
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow
```

### Delete the element

```python
def delete_element(head: Node, value: int) -> Node:
    dummy = Node(0)
    dummy.next = head
    prev = dummy
    current = head

    while current:
        if current.val == value:
            prev.next = current.next
        else:
            prev = current

        current = current.next
        
    return dummy.next
```

### Remove N-th from the back

```python
def delete_from_back(head: Node, n: int) -> Node:
    dummy = Node(0, head)
    slow = dummy
    fast = dummy

    for _ in range(n + 1):
        fast = fast.next

    while fast:
        fast = fast.next
        slow = slow.next

    slow.next = slow.next.next
    return dummy.next
```

### Is it subsequence?

```python
def isSubsequence(first: str, second: str) -> bool:
    q = deque(first)

    for elements in second:
        if q and q[0] == elements:
            q.popleft()

    return len(q) == 0
```

```python
isSubsequence('abd', 'uabqd')
```

```python
isSubsequence('abdr', 'uabqd')
```

### Is palindrome?

```python
def isPalindrome(a: str) -> bool:
    stack = []
    for char in a:
        stack.append(char)

    for char in a:
        if char != stack.pop():
            return False

    return True
```

```python
isPalindrome('abracadabra')
```

```python
isPalindrome('abracarba')
```

### Stack sorting with stack

```python
def sorting(array: list) -> list:
    stack = []

    while len(array) > 0:
        current = array.pop()

        while len(stack) > 0 and stack[-1] > current:
            array.append(stack.pop())

        stack.append(current)

    return stack
```

```python
sorting([8, 9, 0, 6, 6, 1, 11, 4])
```

### Stack 2 sorted lists

```python
def merge(list1: Node, list2: Node) -> Node:
    dummy = Node(0)
    current = dummy

    while list1 is not None and list2 is not None:
        if list1.val < list2.val:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next

        current = current.next

    if list1 is not None:
        current.next = list1
    else:
        current.next = list2

    return dummy.next
```
