```python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

### Building tree

```python
def build_tree(arr: list[int]) -> TreeNode:
    if not arr or arr[0] is None:
        return None
        
    root = TreeNode(arr[0])
    queue = deque([root])
    i = 1
    
    while queue and i < len(arr):
        node = queue.popleft()
        
        if i < len(arr) and arr[i] is not None:
            node.left = TreeNode(arr[i])
            queue.append(node.left)
        i += 1
        
        if i < len(arr) and arr[i] is not None:
            node.right = TreeNode(arr[i])
            queue.append(node.right)
        i += 1
        
    return root
```

Для работы с деревом используем очередь для поддержания порядка обрбатываемых вершин. Достаем корень, следующие 2 элемента встраиваются за ним и имеют индексы 2i+1 и 2i+2

### Symmetric Binary Tree

```python
def is_symmetric(root: TreeNode) -> bool:

    def checking_parts(part1: TreeNode, part2: TreeNode) -> bool:
        if not part1 and not part2:
            return True
        if not part1 or not part2:
            return False

        return (part1.val == part2.val and checking_parts(part1.left, part2.right) and checking_parts(part1.right, part2.left))

    if root: 
        return checking_parts(root, root)
    else:
        return True         
```

В рекурсии сравниваем все компоненты правого и левого поддеревьев (значения val, а также дальше потомков из вершины запуска, но зеркально). Корректность можно показать тем, что алгоритм проходит по всему дереву полностью, условие на симметричность всего дерева дает нам гарантию существования зеркально отраженной подчасти дерева. 

### Minimal depth search

```python
def min_depth(root: TreeNode) -> int:
    if not root:
        return 0
        
    queue = deque([(root, 1)])
    
    while queue:
        node, depth = queue.popleft()
        
        if not node.left and not node.right:
            return depth
            
        if node.left:
            queue.append((node.left, depth + 1))
        if node.right:
            queue.append((node.right, depth + 1))
            
    return 0
```

BFS применяется, тк идет по дереву послойно. При использовании DFS, мы грубо говоря "ныряем" в самый низ дерева, что в данном случае было бы избыточно. Нам нужно найти первый минимум. 

### Min max product

```python
def min_max_product_bst(root: TreeNode) -> int:
    if not root:
        return None
        
    min_node = root
    while min_node.left:
        min_node = min_node.left
        
    max_node = root
    while max_node.right:
        max_node = max_node.right
        
    return min_node.val * max_node.val
```

Значения, которые мы ищем находятся в листьях дерева, значит нужно пройти всю высоту, те O(h), чтобы до них дойти

### Binary tree comparison

```python
def tree_compare(root1: TreeNode, root2: TreeNode) -> bool:
    if not root1 and not root2:
        return True
        
    if not root1 or not root2:
        return False
    if root1.val != root2.val:
        return False
        
    return tree_compare(root1.left, root2.left) and tree_compare(root1.right, root2.right)
```

Деревья совпадают при полном совпадении структуры. На каждом условии проверяем именно это.
