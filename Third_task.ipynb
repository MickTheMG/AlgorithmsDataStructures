### Root of the number

```python
def my_sqrt(n):
    if n < 0:
        raise ValueError('sqrt must be > 0')
    if n < 2:
        return n
    left = 1
    right = n // 2
    res = 0

    while left <= right:
        mid = (left + right) // 2
        square = mid * mid 

        if square == n:
            return mid
        elif square < n:
            res = mid
            left = mid + 1
        else:
            right = mid - 1
    return res
```

```python
my_sqrt(36)
```

### Two printers

```python
def printers(x, y, n):
    if n == 1:
        return min(x, y)
    time = 0
    time += min(x, y)

    left = 0
    right = min(x, y) * (n - 1)

    remaining_copies = n - 1
    best_time = right

    while left <= right:
        mid = (right + left) // 2

        made =  mid // x  + mid // y
        if made >= remaining_copies:
            best_time = mid
            right = mid - 1
        else:
            left = mid + 1
    return time + best_time
```

```python
printers(3,3,4)
```

### Feed the animals

```python
def feed(animals, food):
    animals.sort()
    food.sort()

    animal_p, food_p, fed_cnt = 0, 0, 0
    while animal_p < len(animals) and food_p < len(food):
        if animals[animal_p] <= food[food_p]:
            fed_cnt += 1
            animal_p += 1
            food_p += 1
        else:
            food_p += 1
    return fed_cnt
```

```python
animals = [3, 5, 2, 8]
food = [1, 4, 6]
feed(animals, food)
```

### Difference between the strings

```python
def diff(a, b):
    char_counts = {}
    for char in a:
        if char in char_counts:
            char_counts[char] += 1
        else:
            char_counts[char] = 1

    for char in b:
        if char not in char_counts or char_counts[char] == 0:
            return char
        char_counts[char] -= 1

    
```

```python
a = 'aa'
b = 'ab'
print (diff(a, b))
```

### Summ of two elements

```python
def two_sum(nums, target):
    seen = {}
    for i, cur in enumerate(nums):
        residual = target - cur
        if residual not in seen:
            seen[cur] = i
        else:
            return [seen[residual], i]
    return []
```

```python
nums = [1, 3, 2, 8, 4]
target = 10

two_sum(nums, target)
```

### Anogramm array

```python
def anagramm_array(ar):
    dct = {}
    for word in ar:
        word_sorted = "".join(sorted(word))
        dct.setdefault(word_sorted, [])
        dct[word_sorted].append(word)
    return list(dct.values())
```

```python
ar = ["eat","tea","tan","ate","nat","bat"]
anagramm_array(ar)
```
