```python
from typing import List
import math
```

### Two summ

```python
def two_summ (array: List[int], target: int) -> List[int]:
    left, right = 0, len(array) - 1

    while left < right:
        ans = array[left] + array[right]

        if ans == target:
            return[left, right]
        elif ans < target:
            left += 1
        else:
            right -= 1

    return []
```

```python
two_summ([3, 8, 9, 11, 16, 18, 19, 21], 25)
```

### Reverse array

```python
def reverse_array(array: List[int]) -> List[int]:
    left, right = 0, len(array) - 1

    while left < right:
        array[left], array[right] = array[right], array[left]
        left += 1
        right -= 1

    return array
```

```python
reverse_array([1, 2, 3, 4])
```

### Reverse part of array

```python
def reverse_subarray(array: List[int], left: int, right: int) -> None:
    while left < right:
        array[left], array[right] = array[right], array[left]
        left += 1
        right -= 1

def rotation(array: List[int], position: int) -> List[int]:
    length = len(array)
    position = position % length

    reverse_subarray(array, 0, length - 1)
    reverse_subarray(array, 0, position - 1)
    reverse_subarray(array, position, length - 1)

    return array
```

```python
rotation([1, 2, 3, 4, 5, 6, 7], 8)
```

### Merge two sorted arrays

```python
def merge_sorted_arrays(array1: List[int], array2: List[int]) -> List[int]:
    merged_array = []
    i, j = 0, 0

    length1 = len(array1)
    length2 = len(array2)

    while i < length1 and j < length2:
        if array1[i] < array2[j]:
            merged_array.append(array1[i])
            i+=1
        else:
            merged_array.append(array2[j])
            j+=1

    merged_array.extend(array1[i:])
    merged_array.extend(array2[j:])

    return merged_array
        
```

```python
merge_sorted_arrays([2, 4, 6, 8], [1, 3, 4, 4, 11])
```

### Merge optimisation (without new array)

```python
def merge_optimised(array1: List[int], array2: List[int]) -> List[int]:
    length1 = len(array1)
    length2 = len(array2)

    pointer_both = length1 - length2 - 1
    pointer1 = length1 - 1
    pointer2 = length2 - 1
    
    while pointer2 >= 0:
        if pointer_both >= 0 and array1[pointer_both] > array2[pointer2]:
            array1[pointer1] = array1[pointer_both]
            pointer_both -= 1
        else:
            array1[pointer1] = array2[pointer2]
            pointer2 -= 1

        pointer1 -= 1

    return array1
```

```python
merge_optimised([3, 8, 10, 11, 0, 0, 0], [1, 7, 9])
```

### Minimal subarray size

```python
def min_subarray(array: List[int], target: int) -> int:
    min_length = math.inf
    left = 0
    current_summ = 0

    for right in range(len(array)):
        current_summ += array[right]

        while current_summ >= target:
            window_size = right - left + 1

            if window_size < min_length:
                min_length = window_size

            current_summ -= array[left]
            left += 1

    return min_length if min_length != math.inf else 0
```

```python
min_subarray([2, 3, 1, 2, 4, 3], 9)
```

### Sort 0 and 1

```python
def sort(array: List[int]) -> List[int]:
    left, right = 0, len(array) - 1

    while left < right:
        if array[left] == 0:
            left += 1
        elif array[right] == 1:
            right -= 1
        else:
            array[left], array[right] = array[right], array[left]
            left += 1
            right -= 1

    return array
```

```python
sort([0, 1, 1, 0, 0, 1, 0, 0, 1, 1])
```

### Colors sort

```python
def sort_color(array: List[int]) -> List[int]:
    low, mid, high = 0, 0, len(array) - 1

    while mid <= high:
        if array[mid] == 0:
            array[low], array[mid] = array[mid], array[low]
            low += 1
            mid -= 1
        elif array[mid] == 1:
            mid += 1
        elif array[mid] == 2:
            array[mid], array[high] = array[high], array[mid]
            high -= 1

    return array
```

```python
sort_color([1, 2, 0, 0, 1, 1, 1, 0, 2, 2, 2, 1])
```

### Move even numbers

```python
def move_even(array: List[int]) -> List[int]:
    index_of_even = 0

    for i in range (len(array)):
        if array[i] % 2 == 0:
            array[i], array[index_of_even] = array[index_of_even], array[i]
            index_of_even += 1

    return array
```

```python
move_even([1, 1, 1, 2, 3, 3, 4, 4, 4, 8, 7, 5, 0])
```

### 0 in the end

```python
def move_zero(array: List[int]) -> List[int]:
    no_zero_index = 0

    for i in range(len(array)):
        if array[i] != 0:
            array[i], array[no_zero_index] = array[no_zero_index], array[i]
            no_zero_index += 1

    return array
```

```python
move_zero([0, 0, 1, 0, 3, 12])
```

```python
move_zero([0, 0, 0, 18, 16, 0, 0, 77, 99])
```
